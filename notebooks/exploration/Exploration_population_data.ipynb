```python
df = df.rename(columns={'Time': 'YEAR'})
```

```python
print(df['YEAR'].min(), df['YEAR'].max())
```

```python
df = df[(df['YEAR'] >= 1994) & (df['YEAR'] <= 2024)].reset_index(drop=True)
```

```python
df['ISO3_code'].nunique()
```

```python
df['ISO3_code']
```

```python
bcg = pd.read_parquet('../../data/processed/bcg_doses_clean.parquet')
pop_countries = set(df['ISO3_code'].unique())
bcg_countries = set(bcg['CODE'].unique())

only_in_pop = pop_countries - bcg_countries
only_in_bcg = bcg_countries - pop_countries

print(f"Countries in population only: {len(only_in_pop)}")
print(f"Countries in BCG only: {len(only_in_bcg)}")
print(f"Countries in both: {len(pop_countries & bcg_countries)}")
```

```python
merged = bcg.merge(
    df.rename(columns={'ISO3_code': 'CODE'}),
    on=['CODE', 'YEAR'],
    how='inner'
)
print(merged.shape)
merged.head()
```
